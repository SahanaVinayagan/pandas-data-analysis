```python
import pandas as pd
from IPython.display import display

#Load dataset
df = pd.read_csv("student_performance.csv")

# Display first 5 rows
display(df.head())
```

```python
# Display last 5 rows
display(df.tail())
```

```python
# Display number of rows and columns
print(df.shape)
```

```python
# Display the column names
print(df.columns)
```

```python
# Print datatypes
print(df.dtypes)
```

```python
# Display information about the DataFrame
print(df.info())
```

```python
# Display summary statistics of numerical columns
display(df.describe())
```

```python
# Check the number of missing values in each column
df.isnull().sum()
```

```python
# Check whether any missing value exists in each column
df.isnull().any()
```

```python
# Find columns containing missing values
null_columns = df.columns[df.isnull().any()]
null_columns
```

```python
# Handle Missing Values

# Fill missing values in a numerical column with the mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Fill missing values in a numerical column with the median
df['Final_GRade'] = df['Final_Grade'].fillna(df['Final_Grade'].median())

# Fill missing values in a categorical column with the mode
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
# Check duplicate rows
print(df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

print(df.duplicated().sum())
```

```python
# Select a single column
df['Age']
```

```python
# Select multiple columns
df[['Age', 'Gender', 'Final_Grade']]
```

```python
# Select the first row using its label
df.loc[0]
```

```python
# Select the first five rows using their integer positions
df.iloc[0:5]
```

```python
# Select the Name column for the first five rows
df.loc[0:4, 'Student_ID']
```

```python
# Filtering data

high_score = df[df['Final_Grade'] > 80]
display(high_score)

display(df[(df['Attendance_Percentage'] > 95) & (df['Study_Hours_per_Week'] > 15)])
```

```python
# Sort the data

# Ascending order
display(df.sort_values('Age'))

# Descending order
df.sort_values('Projects_Submitted', ascending=False)
```

```python
#Find unique values in Teaching_Effectiveness column
display(df['Teaching_Effectiveness'].unique())

# Count the number of unique values in Gender column
display(df['Gender'].nunique())

# Count the frequency of each value in the Student_Performance column
display(df['Student_Performance'].value_counts())
```

```python
# Basic mathematical operations

print("Mean:",df['Midterm_Grade'].mean())
print("Median:",df['Midterm_Grade'].median())

# Mode for numerical data
print("Mode:",df['Midterm_Grade'].mode()[0])

# Mode for categorical data
print("Mode:",df['Gender'].mode()[0])

print("Max:",df['Midterm_Grade'].max())
print("Min:",df['Midterm_Grade'].min())
print("Sum:",df['Midterm_Grade'].sum())
print("Standard Deviation:",df['Midterm_Grade'].std())
print("Variance:", df['Midterm_Grade'].var())
print("Range:", df['Midterm_Grade'].max() - df['Midterm_Grade'].min())
print("Count:", df['Midterm_Grade'].count())
```

```python
# Calculate the mean, minimum, and maximum midterm grade for each teaching effectiveness level
df.groupby('Teaching_Effectiveness')['Midterm_Grade'].agg(['mean', 'min', 'max'])
```

```python
# Create a new column
df['Total_Grade'] = df['Midterm_Grade'] + df['Final_Grade']
df.head()
```

```python
# Rename column
df.rename(columns={'Attendance_Percentage': 'Attendance'}, inplace=True)
df.columns.tolist()
```
