```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv("student_performance.csv")
```

```python
# Bar Graph
# Shows the number of male and female students 

df['Gender'].value_counts().plot(kind='bar')

plt.title('Number of Students by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Students')
plt.show()
```

```python
# Histogram
# Shows how attendance is distributed among the students

df['Attendance_Percentage'].plot(kind='hist', bins=10, color="purple")

plt.title('Distribution of Attendance Percentage')
plt.xlabel('Attendance Percentage')
plt.ylabel('Number of Students')
plt.show()
```

```python
# Pie Chart
# Shows the information as percentage 

df['Gender'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Gender Distribution')
plt.show()
```

```python
# Box Plot

df['Final_Grade'].plot(kind='box')

plt.title('Final Grade Distribution')
plt.ylabel('Final Grade')
plt.show()
```

```python
# Scatter Plot

df.plot(
    x='Attendance_Percentage',
    y='Final_Grade',
    kind='scatter'
)

plt.title('Attendance Percentage vs Final Grade')
plt.xlabel('Attendance Percentage')
plt.ylabel('Final Grade')
plt.show()
```

```python
# Line Plot

df.groupby('Study_Hours_per_Week')['Final_Grade'].mean().plot(kind='line')

plt.title('Average Final Grade by Study Hours')
plt.xlabel('Study Hours per Week')
plt.ylabel('Average Final Grade')
plt.show()
```

```python
# Area Plot

df.groupby('Age')['Final_Grade'].mean().plot(kind='area')

plt.title('Average Final Grade by Age')
plt.xlabel('Age')
plt.ylabel('Average Final Grade')
plt.show()
```

```python
# Hexbin Plot

df.plot(
    x='Attendance_Percentage',
    y='Final_Grade',
    kind='hexbin',
    gridsize=15
)

plt.title('Attendance Percentage vs Final Grade')
plt.xlabel('Attendance Percentage')
plt.ylabel('Final Grade')
plt.show()
```

```python
# Density/KDE Plot

df['Final_Grade'].plot(kind='kde')

plt.title('Density Distribution of Final Grades')
plt.xlabel('Final Grade')
plt.ylabel('Density')
plt.show()
```

```python
# Stacked Bar Plot

pd.crosstab(
    df['Teaching_Effectiveness'],
    df['Student_Performance']
).plot(kind='bar', stacked=True)

plt.title('Student Performance by Teaching Effectiveness')
plt.xlabel('Teaching Effectiveness')
plt.ylabel('Number of Students')
plt.show()
```

```python
# Horizontal Bar Plot

df[
    ['Teaching_Clarity',
     'Teaching_Helpfulness',
     'Feedback_Timeliness']
].mean().sort_values().plot(kind='barh')

plt.title('Average Teaching Quality Ratings')
plt.xlabel('Average Rating')
plt.ylabel('Teaching Factor')
plt.show()
```

```python
# Multiple Line Plot

df[
    ['Teaching_Clarity',
     'Teaching_Helpfulness',
     'Feedback_Timeliness']
].mean().plot(kind='line', marker='o')

plt.title('Average Teaching Quality Factors')
plt.xlabel('Teaching Factors')
plt.ylabel('Average Rating')
plt.show()
```

```python
# Density Plot

df.groupby('Gender')['Final_Grade'].plot(kind='kde', legend=True)

plt.title('Final Grade Distribution by Gender')
plt.xlabel('Final Grade')
plt.ylabel('Density')
plt.show()
```

```python
# Violin Plot

import seaborn as sns
import matplotlib.pyplot as plt

sns.violinplot(
    x='Gender',
    y='Final_Grade',
    data=df,
    color="Pink"
)

plt.title('Final Grade Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Final Grade')
plt.show()
```

```python
# Swarm Plot

sns.swarmplot(
    x='Student_Performance',
    y='Final_Grade',
    data=df
)

plt.title('Final Grade by Student Performance')
plt.xlabel('Student Performance')
plt.ylabel('Final Grade')
plt.show()
```

```python
# Strip Plot

sns.stripplot(
    x='Student_Performance',
    y='Final_Grade',
    data=df
)

plt.title('Final Grade by Student Performance')
plt.xlabel('Student Performance')
plt.ylabel('Final Grade')
plt.show()
```

```python
# Count Plot

sns.countplot(
    x='Teaching_Effectiveness',
    data=df
)

plt.title('Teaching Effectiveness Count')
plt.xlabel('Teaching Effectiveness')
plt.ylabel('Number of Students')
plt.show()
```

```python
# Regression Plor

sns.regplot(
    x='Study_Hours_per_Week',
    y='Final_Grade',
    data=df
)

plt.title('Study Hours vs Final Grade')
plt.xlabel('Study Hours per Week')
plt.ylabel('Final Grade')
plt.show()
```

```python
# Joint Plot

sns.jointplot(
    x='Attendance_Percentage',
    y='Final_Grade',
    data=df
)
```

```python
# Pair Plot

sns.pairplot(
    df[
        ['Age',
         'Attendance_Percentage',
         'Midterm_Grade',
         'Final_Grade',
         'Study_Hours_per_Week']
    ]
)

plt.show()
```

```python
# Heatmap

correlation = df.corr(numeric_only=True)

sns.heatmap(
    correlation,
    annot=True
)

plt.title('Correlation Between Numerical Variables')
plt.show()
```

```python
# ECDF Plot

sns.ecdfplot(
    data=df,
    x='Final_Grade'
)

plt.title('Cumulative Distribution of Final Grades')
plt.xlabel('Final Grade')
plt.ylabel('Proportion of Students')
plt.show()
```
